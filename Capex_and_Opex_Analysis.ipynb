```python
"""
=============================================================================
 DAC SYSTEM — CAPEX / OPEX / ROI ANALYSIS
=============================================================================

 This module performs a complete techno-economic analysis of the
 electrochemical DAC system (absorber + EC cell) for installation
 in Pakistan, targeting >99% pure CO₂ via membrane gas separation.


 Revenue target: Food/beverage-grade CO₂ (>99% purity) in Pakistan

=============================================================================
 REFERENCES — All prices sourced March 2026
=============================================================================

 [R1]  Pakistan industrial electricity tariff (B1, FY2025-26):
       Rs. 26.23/kWh flat rate
       Source: NEPRA / ProPakistani (Feb 2026)
       https://propakistani.pk/2026/02/17/pakistan-cuts-industrial-power-tariffs-by-up-to-rs-4-58-per-unit/

 [R2]  USD/PKR exchange rate: 280 PKR/USD (interbank, 31 Mar 2026)
       Source: PakObserver
       https://pakobserver.net/usd-pkr-rate-other-currency-rates-in-pakistan-today-31-march-2026/

 [R3]  NaOH (caustic soda) bulk price: Rs. 222/kg
       Source: Zarea B2B Platform
       https://zarea.com/categories/chemicals-prices-in-pakistan

 [R4]  Nafion 117 membrane: $1,500/m² (mid-range, 2025-26)
       Source: Thunder Said Energy; Labxsci
       https://thundersaidenergy.com/downloads/nafion-membranes-costs-and-hydrogen-crossover/
       https://labxsci.com/product/n117-cation-exchange-membrane/

 [R5]  Platinum spot price: $63/g (March 2026)
       Source: Kitco / MarketsHost
       https://www.kitco.com/charts/platinum
       https://www.marketshost.com/platinum-price/us/

 [R6]  Mellapak 250Y structured packing: $530-730/m³
       Source: Made-in-China supplier listings
       https://aitemt21.en.made-in-china.com/product/anGrwRFOneUd/

 [R7]  SS316 fabrication cost Pakistan: Rs. 950/kg (material + fabrication)
       Source: MWPBNP / Silver Steel Mills
       https://blog.mwpbnp.com/316-stainless-steel-price-per-kg-in-pakistan/
       https://silversteelmills.com/product/steel-fabrication-company-in-pakistan/

 [R8]  HDPE tank (200L): Rs. 3,700-5,000
       Source: CostZone / JMBS Pakistan
       https://costzone.org/water-tank-prices-in-pakistan-october-2025/
       https://jmbs.pk/water-tank-price-in-pakistan/

 [R9]  Pakistan skilled labor wage: Rs. 46,000/month (Punjab, 2025-26)
       Source: ProPakistani / CLR Punjab
       https://propakistani.pk/2025/10/15/punjab-raises-salaries-for-daily-wage-workers/

 [R10] Food-grade CO₂ price Pakistan: Rs. 800-2,500/kg retail
       Import price ~$6,153/ton (2023 data, rising trend)
       Source: IndexBox / Pakistan Oxygen Ltd
       https://www.indexbox.io/search/carbon-dioxide-price-pakistan/
       https://pakoxygen.com/Gases/Carbondioxide

 [R11] CO₂ global food/beverage grade: $200-780/ton regionally
       Source: ChemAnalyst / IMARC Group
       https://www.chemanalyst.com/Pricing-data/liquid-carbon-dioxide-1090
       https://www.imarcgroup.com/carbon-dioxide-pricing-report

 [R12] Carbon credits (voluntary, DAC/CDR): $170-500/ton
       Source: Sylvera / Regreener
       https://www.sylvera.com/blog/carbon-offset-price
       https://www.regreener.earth/blog/voluntary-carbon-market-update

 [R13] Green hydrogen price: $2.50-5.00/kg (Pakistan/MENA region)
       Source: Energy Solutions / BusinessAnalytiq
       https://energy-solutions.co/articles/sub/green-hydrogen-production-costs
       https://businessanalytiq.com/procurementanalytics/index/green-hydrogen-price-index/

 [R14] Pakistan water tariff (industrial): ~Rs. 300/m³ estimated
       Source: WASA Lahore / CDA Islamabad
       https://wasa.punjab.gov.pk/tariff
       https://pakobserver.net/new-water-bill-charges-in-islamabad-implemented-details-inside/

=============================================================================
"""

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from dataclasses import dataclass, field
from typing import Dict, List, Tuple


# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 1: ECONOMIC PARAMETERS & MARKET DATA                       ║
# ╚═══════════════════════════════════════════════════════════════════════╝

@dataclass
class EconomicParams:
    """All economic parameters with references."""

    # --- Exchange Rate [R2] ---
    USD_PKR: float = 280.0  # PKR per USD, interbank Mar 2026

    # --- Electricity [R1] ---
    elec_rate_PKR: float = 26.23       # PKR/kWh, B1 industrial flat
    elec_rate_USD: float = 26.23/280   # USD/kWh

    # --- Chemicals [R3] ---
    NaOH_price_PKR_kg: float = 222.0   # PKR/kg, bulk industrial
    NaOH_price_USD_kg: float = 222/280

    # --- Water [R14] ---
    water_price_PKR_m3: float = 300.0  # PKR/m³, industrial estimate

    # --- Membranes [R4] ---
    nafion_price_USD_m2: float = 1500.0  # $/m², mid-range

    # --- Platinum [R5] ---
    Pt_price_USD_g: float = 63.0  # $/g spot

    # --- Packing [R6] ---
    mellapak_price_USD_m3: float = 730.0  # $/m³ (small order)

    # --- SS316 Fabrication [R7] ---
    SS316_price_PKR_kg: float = 950.0   # PKR/kg all-in

    # --- HDPE Tank [R8] ---
    HDPE_tank_200L_PKR: float = 5000.0  # PKR each

    # --- Labor [R9] ---
    skilled_wage_PKR_mo: float = 46000.0  # PKR/month

    # --- CO₂ Prices [R10, R11] ---
    CO2_food_grade_PKR_kg: float = 1200.0  # PKR/kg Pakistan mid-range
    CO2_food_grade_USD_ton: float = 500.0   # $/ton conservative mid
    CO2_export_USD_ton: float = 320.0       # $/ton Europe benchmark

    # --- Carbon Credits [R12] ---
    carbon_credit_CDR_USD_ton: float = 250.0  # $/ton DAC/CDR credits

    # --- Hydrogen [R13] ---
    H2_price_USD_kg: float = 4.0  # $/kg green H₂ Pakistan estimate

    # --- Financial ---
    discount_rate: float = 0.10    # 10% for Pakistan
    project_life_years: int = 20
    inflation_rate: float = 0.08   # Pakistan CPI ~8%
    import_duty_pct: float = 0.15  # 15% on imported equipment


# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 2: PLANT DESIGN DATA (from simulation models)              ║
# ╚═══════════════════════════════════════════════════════════════════════╝

@dataclass
class PlantDesign:
    """Design parameters extracted from DAC_ecell.txt and absorber.txt."""

    # --- Absorber Column (absorber.txt output) ---
    abs_D_m: float = 0.29           # column diameter [m]
    abs_Z_m: float = 15.0           # packed height [m]
    abs_A_col_m2: float = 0.06605   # cross-section area [m²]
    abs_V_packing_m3: float = 0.0   # computed in __post_init__
    abs_packing_type: str = "Mellapak 250Y (structured)"
    abs_a_p: float = 250.0          # m²/m³
    abs_epsilon: float = 0.97
    abs_G_vol_m3h: float = 35.0     # gas flow [m³/h]
    abs_L_vol_Lmin: float = 40.0    # liquid flow [L/min]
    abs_c_NaOH_in: float = 2.0      # mol/L
    abs_eta_capture: float = 99.66  # % capture efficiency
    abs_CO2_removed_mols: float = 1.691679e-4  # mol/s steady state
    abs_shell_thickness_m: float = 0.003  # 3 mm SS316
    abs_shell_mass_kg: float = 0.0  # computed

    # --- EC Cell (DAC_ecell.txt output) ---
    ec_N_cell: int = 7
    ec_A_cell_cm2: float = 500.0
    ec_A_cell_m2: float = 0.05      # per cell
    ec_A_total_m2: float = 0.35     # total membrane area
    ec_j_applied: float = 92.4     # A/m²
    ec_I_per_cell: float = 4.62    # A
    ec_V_cell_avg: float = 0.7111  # V
    ec_V_stack: float = 4.98       # V
    ec_P_cell_W: float = 23.0
    ec_P_pump_W: float = 1.7
    ec_P_PSU_loss_W: float = 2.0
    ec_P_total_W: float = 26.7
    ec_PSU_rating_W: float = 50.0
    ec_d_mem_um: float = 178       # Nafion 117
    ec_Pt_mass_g: float = 0.20
    ec_stack_mass_dry_kg: float = 30.0
    ec_stack_W_m: float = 0.31     # stack dimensions
    ec_stack_H_m: float = 0.23
    ec_stack_L_m: float = 0.111

    # --- Cycle / Performance (both models) ---
    cycle_time_h: float = 331.1    # hours per cycle
    cycles_per_year: float = 0.0   # computed
    CO2_per_cycle_mol: float = 183.3
    CO2_per_cycle_kg: float = 8.065  # 183.3 × 0.04401
    NaOH_makeup_per_cycle_kg: float = 1.29  # 32.3 mol × 40 g/mol
    H2O_consumed_per_cycle_kg: float = 6.6
    H2_per_cycle_mol: float = 183.7  # cathode byproduct
    H2_per_cycle_kg: float = 0.370
    CO2_recovery_pct: float = 96.5
    NaOH_product_per_cycle_mol: float = 369.4  # recycled back
    SEC_kJ_mol: float = 160.3
    SEC_kWh_ton: float = 1012.0
    energy_per_cycle_kJ: float = 0.0  # computed

    # --- Tank Volumes ---
    V_acid_init_L: float = 199.2
    V_base_init_L: float = 200.6
    V_acid_final_L: float = 171.1
    V_base_final_L: float = 214.5

    # --- Blower (absorber) ---
    blower_power_W: float = 50.0
    abs_dP_Pa: float = 51.2

    def __post_init__(self):
        """Compute derived quantities."""
        self.abs_V_packing_m3 = self.abs_A_col_m2 * self.abs_Z_m
        self.cycles_per_year = (365.25 * 24) / self.cycle_time_h
        self.energy_per_cycle_kJ = self.SEC_kJ_mol * self.CO2_per_cycle_mol

        # Absorber shell mass: cylinder + 2 heads
        D = self.abs_D_m
        Z = self.abs_Z_m
        t = self.abs_shell_thickness_m
        rho_ss = 8000  # kg/m³ SS316
        A_shell = np.pi * D * Z  # cylinder
        A_heads = 2 * np.pi * (D/2)**2  # flat heads (approx)
        self.abs_shell_mass_kg = (A_shell + A_heads) * t * rho_ss


# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 3: CAPEX MODEL                                             ║
# ╚═══════════════════════════════════════════════════════════════════════╝

@dataclass
class CAPEXItem:
    name: str
    category: str
    qty: float
    unit: str
    unit_cost_USD: float
    ref: str
    imported: bool = False

    @property
    def base_cost_USD(self) -> float:
        return self.qty * self.unit_cost_USD


def compute_capex(plant: PlantDesign, econ: EconomicParams) -> Dict:
    """Compute detailed CAPEX breakdown."""
    PKR = econ.USD_PKR
    items = []

    # ── ABSORBER COLUMN ──────────────────────────────────────────────
    # Shell (SS316)
    items.append(CAPEXItem(
        "Absorber Shell (SS316)", "Absorber",
        plant.abs_shell_mass_kg, "kg",
        econ.SS316_price_PKR_kg / PKR,
        "[R7] SS316 @ Rs.950/kg"
    ))

    # Packing
    items.append(CAPEXItem(
        "Mellapak 250Y Packing", "Absorber",
        plant.abs_V_packing_m3, "m³",
        econ.mellapak_price_USD_m3,
        "[R6] Mellapak @ $730/m³",
        imported=True
    ))

    # Liquid distributor
    items.append(CAPEXItem(
        "Liquid Distributor (SS316)", "Absorber",
        1, "unit", 200.0,
        "[R7] SS316 fabrication estimate"
    ))

    # Mist eliminator
    items.append(CAPEXItem(
        "Mist Eliminator (SS mesh)", "Absorber",
        1, "unit", 100.0,
        "[R7] SS316 fabrication estimate"
    ))

    # Support grid & holddown
    items.append(CAPEXItem(
        "Support Grid & Holddown", "Absorber",
        1, "unit", 80.0,
        "[R7] SS316 fabrication estimate"
    ))

    # ── EC CELL STACK ────────────────────────────────────────────────
    # Nafion 117 membranes
    items.append(CAPEXItem(
        "Nafion 117 CEM Membranes", "EC Stack",
        plant.ec_A_total_m2, "m²",
        econ.nafion_price_USD_m2,
        "[R4] Nafion 117 @ $1,500/m²",
        imported=True
    ))

    # Pt catalyst
    items.append(CAPEXItem(
        "Platinum Catalyst (electrodes)", "EC Stack",
        plant.ec_Pt_mass_g, "g",
        econ.Pt_price_USD_g,
        "[R5] Pt @ $63/g spot",
        imported=True
    ))

    # Ti mesh electrode substrates (8 electrodes, ~1.8 kg)
    items.append(CAPEXItem(
        "Ti Mesh Electrode Substrates", "EC Stack",
        1.8, "kg", 15.0,
        "Ti mesh estimate $15/kg",
        imported=True
    ))

    # Bipolar plates (graphite composite)
    items.append(CAPEXItem(
        "Bipolar Plates (graphite)", "EC Stack",
        max(plant.ec_N_cell - 1, 1), "units", 20.0,
        "Graphite composite estimate"
    ))

    # Gaskets & hardware
    items.append(CAPEXItem(
        "Gaskets & Hardware (EPDM/PTFE)", "EC Stack",
        1, "set", 60.0,
        "Sealing materials estimate"
    ))

    # End plates (steel)
    items.append(CAPEXItem(
        "End Plates (steel)", "EC Stack",
        28.0, "kg",
        econ.SS316_price_PKR_kg / PKR,
        "[R7] Steel @ Rs.950/kg"
    ))

    # Stack assembly frame
    items.append(CAPEXItem(
        "Stack Assembly & Frame", "EC Stack",
        1, "unit", 80.0,
        "Fabrication estimate"
    ))

    # ── BALANCE OF PLANT ─────────────────────────────────────────────
    # Tanks (HDPE)
    for name_t in ["Acid Tank (HDPE, 240L)", "Base Tank (HDPE, 240L)",
                    "Absorber Sump Tank (HDPE, 200L)"]:
        items.append(CAPEXItem(
            name_t, "BOP",
            1, "unit",
            econ.HDPE_tank_200L_PKR / PKR,
            "[R8] HDPE tank @ Rs.5,000"
        ))

    # Pumps (3×: 2 EC loop + 1 absorber)
    items.append(CAPEXItem(
        "Mag-drive Pumps (×3)", "BOP",
        3, "units", 150.0,
        "Centrifugal pump estimate"
    ))

    # DC Power Supply
    items.append(CAPEXItem(
        "DC Power Supply (50W)", "BOP",
        1, "unit", 80.0,
        "PSU estimate",
        imported=True
    ))

    # Gas separator (membrane, for >99% CO₂)
    items.append(CAPEXItem(
        "Membrane Gas Separator (>99% CO₂)", "BOP",
        1, "unit", 350.0,
        "Membrane gas separator estimate",
        imported=True
    ))

    # Piping & manifold
    items.append(CAPEXItem(
        "Piping & Manifold (CPVC/PP, ~8m)", "BOP",
        1, "set", 120.0,
        "Piping materials estimate"
    ))

    # Valves, fittings, instruments
    items.append(CAPEXItem(
        "Valves, Fittings, Instruments", "BOP",
        1, "set", 300.0,
        "pH probes, flow meters, gauges"
    ))

    # Air blower
    items.append(CAPEXItem(
        "Air Blower/Fan (35 m³/h)", "BOP",
        1, "unit", 150.0,
        "Centrifugal blower estimate"
    ))

    # Control panel
    items.append(CAPEXItem(
        "Control Panel / Basic PLC", "BOP",
        1, "unit", 250.0,
        "Simple automation estimate"
    ))

    # Wiring & electrical
    items.append(CAPEXItem(
        "Wiring & Electrical", "BOP",
        1, "set", 100.0,
        "Electrical materials"
    ))

    # ── INSTALLATION & CONTINGENCY ───────────────────────────────────
    items.append(CAPEXItem(
        "Site Preparation (concrete pad)", "Install",
        1, "unit", 100.0,
        "Civil works estimate"
    ))

    items.append(CAPEXItem(
        "Installation Labor", "Install",
        1, "unit",
        2 * econ.skilled_wage_PKR_mo * 0.5 / PKR,  # 2 workers × 2 weeks
        "[R9] Skilled labor @ Rs.46,000/mo"
    ))

    items.append(CAPEXItem(
        "Commissioning & Testing", "Install",
        1, "unit", 200.0,
        "Startup estimate"
    ))

    # Compute base subtotal for import duty and contingency
    base_total = sum(it.base_cost_USD for it in items)
    imported_total = sum(it.base_cost_USD for it in items if it.imported)

    items.append(CAPEXItem(
        "Import Duties & Shipping (15% on imports)", "Install",
        imported_total, "USD",
        econ.import_duty_pct,
        "Pakistan customs duty 15%"
    ))

    items.append(CAPEXItem(
        "Contingency (15% of equipment)", "Install",
        base_total, "USD",
        0.15,
        "Standard contingency"
    ))

    # Build summary
    total_USD = sum(it.base_cost_USD for it in items)

    categories = {}
    for it in items:
        cat = it.category
        if cat not in categories:
            categories[cat] = []
        categories[cat].append(it)

    return {
        'items': items,
        'categories': categories,
        'total_USD': total_USD,
        'total_PKR': total_USD * PKR,
    }


# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 4: OPEX MODEL                                              ║
# ╚═══════════════════════════════════════════════════════════════════════╝

@dataclass
class OPEXItem:
    name: str
    category: str
    annual_USD: float
    ref: str
    detail: str = ""


def compute_opex(plant: PlantDesign, econ: EconomicParams,
                 capex_total_USD: float) -> Dict:
    """Compute detailed annual OPEX breakdown."""
    PKR = econ.USD_PKR
    cy = plant.cycles_per_year
    items = []

    # ── ELECTRICITY ──────────────────────────────────────────────────
    # Operating hours per year (both units run ~simultaneously)
    op_hours = cy * plant.cycle_time_h

    elec_items = [
        ("EC Cell", plant.ec_P_cell_W, op_hours),
        ("EC Pumps (×2)", plant.ec_P_pump_W, op_hours),
        ("PSU Losses", plant.ec_P_PSU_loss_W, op_hours),
        ("Absorber Blower/Fan", plant.blower_power_W, op_hours),
        ("Absorber Pump", 10.0, op_hours),  # ~10W for absorber recirculation
        ("Instrumentation", 10.0, 8760),    # always on
    ]

    total_elec_kWh = 0
    for name, W, hrs in elec_items:
        kWh = W * hrs / 1000.0
        cost_USD = kWh * econ.elec_rate_PKR / PKR
        total_elec_kWh += kWh
        items.append(OPEXItem(
            name, "Electricity", cost_USD,
            "[R1] Rs.26.23/kWh",
            f"{W:.0f}W × {hrs:.0f}h = {kWh:.1f} kWh/yr"
        ))

    # ── CHEMICALS ────────────────────────────────────────────────────
    # NaOH makeup
    naoh_annual_kg = plant.NaOH_makeup_per_cycle_kg * cy
    items.append(OPEXItem(
        "NaOH Makeup", "Chemicals",
        naoh_annual_kg * econ.NaOH_price_PKR_kg / PKR,
        "[R3] NaOH @ Rs.222/kg bulk",
        f"{naoh_annual_kg:.1f} kg/yr"
    ))

    # Water
    water_annual_m3 = plant.H2O_consumed_per_cycle_kg * cy / 1000.0
    items.append(OPEXItem(
        "DI Water", "Chemicals",
        water_annual_m3 * econ.water_price_PKR_m3 / PKR,
        "[R14] Water @ Rs.300/m³",
        f"{water_annual_m3:.3f} m³/yr"
    ))

    # Initial NaOH charge amortized (5 yr)
    naoh_charge_kg = plant.abs_c_NaOH_in * plant.V_acid_init_L * 0.040  # mol × MW
    items.append(OPEXItem(
        "NaOH Initial Charge (amortized/5yr)", "Chemicals",
        (naoh_charge_kg / 5) * econ.NaOH_price_PKR_kg / PKR,
        "[R3] NaOH charge amortized",
        f"{naoh_charge_kg:.1f} kg ÷ 5 yr"
    ))

    # ── MEMBRANE & ELECTRODE REPLACEMENT ─────────────────────────────
    nafion_cost = plant.ec_A_total_m2 * econ.nafion_price_USD_m2
    items.append(OPEXItem(
        "Nafion 117 Replacement", "Replacement",
        nafion_cost / 3.0,  # 3-year life
        "[R4] Nafion 117 @ $1,500/m², 3yr life",
        f"${nafion_cost:.0f} ÷ 3 yr"
    ))

    items.append(OPEXItem(
        "Pt Electrode Recoating", "Replacement",
        (plant.ec_Pt_mass_g * econ.Pt_price_USD_g * 0.5) / 5.0,  # 50% Pt reused, 5yr
        "[R5] Pt recoating every 5yr, 50% reuse",
        f"{plant.ec_Pt_mass_g:.2f}g Pt"
    ))

    items.append(OPEXItem(
        "Gaskets & Seals", "Replacement",
        60.0 / 2.0,  # $60 set, 2yr life
        "Gasket set, 2yr life"
    ))

    packing_cost = plant.abs_V_packing_m3 * econ.mellapak_price_USD_m3
    items.append(OPEXItem(
        "Absorber Packing Replacement", "Replacement",
        packing_cost / 10.0,  # 10yr life
        "[R6] Packing @ $730/m³, 10yr life",
        f"${packing_cost:.0f} ÷ 10 yr"
    ))

    # ── LABOR ────────────────────────────────────────────────────────
    items.append(OPEXItem(
        "Part-time Operator (0.25 FTE)", "Labor",
        0.25 * 12 * econ.skilled_wage_PKR_mo / PKR,
        "[R9] Skilled @ Rs.46,000/mo",
        "Monitoring, sampling, dosing"
    ))

    items.append(OPEXItem(
        "Maintenance Tech (0.10 FTE)", "Labor",
        0.10 * 12 * econ.skilled_wage_PKR_mo / PKR,
        "[R9] Skilled @ Rs.46,000/mo",
        "Quarterly inspection"
    ))

    # ── MAINTENANCE & MISC ───────────────────────────────────────────
    items.append(OPEXItem(
        "General Maintenance (3% CAPEX)", "Maintenance",
        0.03 * capex_total_USD,
        "Standard 3% of CAPEX/yr"
    ))

    items.append(OPEXItem(
        "Insurance (1% CAPEX)", "Maintenance",
        0.01 * capex_total_USD,
        "Standard 1% of CAPEX/yr"
    ))

    items.append(OPEXItem(
        "Misc. (sampling, safety, admin)", "Maintenance",
        50.0,
        "Lump sum estimate"
    ))

    total_USD = sum(it.annual_USD for it in items)

    categories = {}
    for it in items:
        cat = it.category
        if cat not in categories:
            categories[cat] = []
        categories[cat].append(it)

    return {
        'items': items,
        'categories': categories,
        'total_USD': total_USD,
        'total_PKR': total_USD * PKR,
        'total_elec_kWh': total_elec_kWh,
    }


# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 5: REVENUE MODEL                                           ║
# ╚═══════════════════════════════════════════════════════════════════════╝

@dataclass
class RevenueScenario:
    name: str
    CO2_price_USD_ton: float
    H2_price_USD_kg: float
    carbon_credit_USD_ton: float
    description: str
    ref: str


def build_revenue_scenarios(econ: EconomicParams) -> List[RevenueScenario]:
    """Define revenue scenarios for >99% pure CO₂."""
    return [
        RevenueScenario(
            "Scenario A: Food-Grade CO₂ (Pakistan domestic)",
            CO2_price_USD_ton=econ.CO2_food_grade_PKR_kg * 1000 / econ.USD_PKR,
            # Rs.1200/kg × 1000 kg/ton ÷ 280 PKR/USD ≈ $4,286/ton
            H2_price_USD_kg=econ.H2_price_USD_kg,
            carbon_credit_USD_ton=0,
            description=(
                "Sell >99% pure CO₂ as food/beverage grade in Pakistan.\n"
                "Pakistan CO₂ retail: Rs.800-2,500/kg → using Rs.1,200/kg mid.\n"
                "H₂ byproduct sold at $4/kg."
            ),
            ref="[R10] IndexBox/PakOxygen; [R13] H₂"
        ),
        RevenueScenario(
            "Scenario B: CO₂ Export (MENA/regional)",
            CO2_price_USD_ton=econ.CO2_export_USD_ton,  # $320/ton
            H2_price_USD_kg=econ.H2_price_USD_kg,
            carbon_credit_USD_ton=0,
            description=(
                "Export food-grade CO₂ at European benchmark ($320/ton).\n"
                "H₂ byproduct sold at $4/kg."
            ),
            ref="[R11] ChemAnalyst Europe benchmark"
        ),
        RevenueScenario(
            "Scenario C: Carbon Credits (DAC/CDR)",
            CO2_price_USD_ton=0,
            H2_price_USD_kg=econ.H2_price_USD_kg,
            carbon_credit_USD_ton=econ.carbon_credit_CDR_USD_ton,  # $250/ton
            description=(
                "Revenue from engineered CDR carbon credits ($250/ton).\n"
                "H₂ byproduct sold at $4/kg.\n"
                "DAC credits trade at $170-500/ton (Sylvera 2026)."
            ),
            ref="[R12] Sylvera/Regreener CDR credits"
        ),
        RevenueScenario(
            "Scenario D: Combined (Pakistan CO₂ + Credits)",
            CO2_price_USD_ton=econ.CO2_food_grade_PKR_kg * 1000 / econ.USD_PKR,
            H2_price_USD_kg=econ.H2_price_USD_kg,
            carbon_credit_USD_ton=econ.carbon_credit_CDR_USD_ton * 0.5,
            # 50% of credits (conservative stacking)
            description=(
                "Domestic food-grade CO₂ sale + 50% carbon credit stacking.\n"
                "Conservative combined revenue."
            ),
            ref="[R10][R12] Combined"
        ),
    ]


def compute_annual_revenue(plant: PlantDesign, scenario: RevenueScenario) -> Dict:
    """Compute annual revenue for a scenario."""
    cy = plant.cycles_per_year
    CO2_annual_ton = plant.CO2_per_cycle_kg * cy / 1000.0
    H2_annual_kg = plant.H2_per_cycle_kg * cy

    rev_CO2 = CO2_annual_ton * scenario.CO2_price_USD_ton
    rev_H2 = H2_annual_kg * scenario.H2_price_USD_kg
    rev_credits = CO2_annual_ton * scenario.carbon_credit_USD_ton

    total = rev_CO2 + rev_H2 + rev_credits

    return {
        'CO2_annual_ton': CO2_annual_ton,
        'H2_annual_kg': H2_annual_kg,
        'rev_CO2': rev_CO2,
        'rev_H2': rev_H2,
        'rev_credits': rev_credits,
        'total': total,
    }


# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 6: ROI & FINANCIAL ANALYSIS                                 ║
# ╚═══════════════════════════════════════════════════════════════════════╝

def compute_roi(capex_USD: float, opex_annual_USD: float,
                revenue_annual_USD: float, econ: EconomicParams) -> Dict:
    """Compute NPV, IRR, payback, and cumulative cash flows."""
    n = econ.project_life_years
    r = econ.discount_rate
    inf = econ.inflation_rate

    years = np.arange(0, n + 1)
    capex_flow = np.zeros(n + 1)
    capex_flow[0] = -capex_USD

    opex_flow = np.zeros(n + 1)
    revenue_flow = np.zeros(n + 1)
    net_flow = np.zeros(n + 1)
    net_flow[0] = -capex_USD
    cum_flow = np.zeros(n + 1)
    cum_flow[0] = -capex_USD
    npv_flow = np.zeros(n + 1)
    npv_flow[0] = -capex_USD

    for yr in range(1, n + 1):
        # Apply inflation to OPEX and revenue
        opex_yr = opex_annual_USD * (1 + inf) ** (yr - 1)
        rev_yr = revenue_annual_USD * (1 + inf) ** (yr - 1)
        opex_flow[yr] = -opex_yr
        revenue_flow[yr] = rev_yr
        net_flow[yr] = rev_yr - opex_yr
        cum_flow[yr] = cum_flow[yr - 1] + net_flow[yr]
        npv_flow[yr] = net_flow[yr] / (1 + r) ** yr

    NPV = np.sum(npv_flow)

    # Payback period (simple)
    payback = None
    for yr in range(1, n + 1):
        if cum_flow[yr] >= 0:
            # Linear interpolation
            frac = -cum_flow[yr - 1] / (cum_flow[yr] - cum_flow[yr - 1])
            payback = (yr - 1) + frac
            break

    # IRR (by bisection)
    def npv_at_rate(rate):
        val = -capex_USD
        for yr in range(1, n + 1):
            opex_yr = opex_annual_USD * (1 + inf) ** (yr - 1)
            rev_yr = revenue_annual_USD * (1 + inf) ** (yr - 1)
            val += (rev_yr - opex_yr) / (1 + rate) ** yr
        return val

    IRR = None
    if revenue_annual_USD > opex_annual_USD:
        try:
            from scipy.optimize import brentq
            IRR = brentq(npv_at_rate, -0.5, 10.0, xtol=1e-6)
        except Exception:
            IRR = None

    # Levelized cost
    annual_CO2_ton = 0.213  # from plant design
    levelized_cost = (capex_USD / n + opex_annual_USD) / annual_CO2_ton

    return {
        'years': years,
        'capex_flow': capex_flow,
        'opex_flow': opex_flow,
        'revenue_flow': revenue_flow,
        'net_flow': net_flow,
        'cum_flow': cum_flow,
        'npv_flow': npv_flow,
        'NPV': NPV,
        'IRR': IRR,
        'payback_years': payback,
        'levelized_cost': levelized_cost,
    }


# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 7: REPORTING & VISUALIZATION                               ║
# ╚═══════════════════════════════════════════════════════════════════════╝

def print_capex_report(capex: Dict, econ: EconomicParams):
    PKR = econ.USD_PKR
    print(f"\n{'='*78}")
    print(f"  CAPEX — Capital Expenditure Breakdown")
    print(f"  Location: Pakistan | Exchange Rate: 1 USD = {PKR:.0f} PKR")
    print(f"{'='*78}")

    for cat_name, cat_items in capex['categories'].items():
        cat_total = sum(it.base_cost_USD for it in cat_items)
        print(f"\n  ── {cat_name} (${cat_total:,.0f} | Rs.{cat_total*PKR:,.0f}) ──")
        print(f"  {'Item':<45s} {'Qty':>6s} {'Unit':>5s} "
              f"{'$/unit':>8s} {'$ Total':>9s} {'Rs.':>12s}  Ref")
        print(f"  {'─'*110}")
        for it in cat_items:
            cost = it.base_cost_USD
            imp = " [IMP]" if it.imported else ""
            print(f"  {it.name:<45s} {it.qty:>6.2f} {it.unit:>5s} "
                  f"${it.unit_cost_USD:>7.1f} ${cost:>8.0f} "
                  f"Rs.{cost*PKR:>10,.0f}  {it.ref}{imp}")

    print(f"\n  {'━'*78}")
    print(f"  TOTAL CAPEX: ${capex['total_USD']:,.0f} "
          f"(Rs. {capex['total_PKR']:,.0f} = "
          f"Rs. {capex['total_PKR']/1e5:.2f} Lakh)")
    print(f"  {'━'*78}")


def print_opex_report(opex: Dict, econ: EconomicParams, plant: PlantDesign):
    PKR = econ.USD_PKR
    print(f"\n{'='*78}")
    print(f"  OPEX — Annual Operating Expenditure")
    print(f"  Cycles/year: {plant.cycles_per_year:.1f} | "
          f"CO₂/year: {plant.CO2_per_cycle_kg * plant.cycles_per_year:.1f} kg")
    print(f"{'='*78}")

    for cat_name, cat_items in opex['categories'].items():
        cat_total = sum(it.annual_USD for it in cat_items)
        print(f"\n  ── {cat_name} (${cat_total:,.0f}/yr | "
              f"Rs.{cat_total*PKR:,.0f}/yr) ──")
        print(f"  {'Item':<42s} {'$/yr':>9s} {'Rs./yr':>12s}  "
              f"{'Detail':<30s} Ref")
        print(f"  {'─'*115}")
        for it in cat_items:
            print(f"  {it.name:<42s} ${it.annual_USD:>8.0f} "
                  f"Rs.{it.annual_USD*PKR:>10,.0f}  "
                  f"{it.detail:<30s} {it.ref}")

    print(f"\n  {'━'*78}")
    print(f"  TOTAL OPEX: ${opex['total_USD']:,.0f}/yr "
          f"(Rs. {opex['total_PKR']:,.0f}/yr = "
          f"Rs. {opex['total_PKR']/1e5:.2f} Lakh/yr)")
    print(f"  Electricity: {opex['total_elec_kWh']:.0f} kWh/yr")
    print(f"  {'━'*78}")


def print_revenue_report(scenarios: List, plant: PlantDesign,
                         econ: EconomicParams):
    PKR = econ.USD_PKR
    print(f"\n{'='*78}")
    print(f"  REVENUE ANALYSIS — >99% Pure CO₂ (membrane gas separator)")
    print(f"  Annual production: {plant.CO2_per_cycle_kg * plant.cycles_per_year:.1f} kg CO₂ "
          f"+ {plant.H2_per_cycle_kg * plant.cycles_per_year:.1f} kg H₂")
    print(f"{'='*78}")

    for sc in scenarios:
        rev = compute_annual_revenue(plant, sc)
        print(f"\n  ── {sc.name} ──")
        print(f"  {sc.description}")
        print(f"  Ref: {sc.ref}")
        print(f"    CO₂ revenue:  ${rev['rev_CO2']:>8.0f}/yr "
              f"(@ ${sc.CO2_price_USD_ton:.0f}/ton)")
        print(f"    H₂ revenue:   ${rev['rev_H2']:>8.0f}/yr "
              f"(@ ${sc.H2_price_USD_kg:.1f}/kg)")
        print(f"    Credit rev:   ${rev['rev_credits']:>8.0f}/yr "
              f"(@ ${sc.carbon_credit_USD_ton:.0f}/ton)")
        print(f"    ────────────────────────")
        print(f"    TOTAL:        ${rev['total']:>8.0f}/yr "
              f"(Rs. {rev['total']*PKR:,.0f}/yr)")


def print_roi_summary(capex: Dict, opex: Dict,
                      scenarios: List, plant: PlantDesign,
                      econ: EconomicParams):
    PKR = econ.USD_PKR
    print(f"\n{'='*78}")
    print(f"  ROI SUMMARY — All Scenarios")
    print(f"  CAPEX: ${capex['total_USD']:,.0f} | OPEX: ${opex['total_USD']:,.0f}/yr | "
          f"Discount: {econ.discount_rate:.0%} | Inflation: {econ.inflation_rate:.0%}")
    print(f"{'='*78}")
    print(f"  {'Scenario':<50s} {'Rev $/yr':>8s} {'NPV $':>10s} "
          f"{'IRR':>7s} {'Payback':>8s} {'LCOC':>10s}")
    print(f"  {'─'*100}")

    for sc in scenarios:
        rev = compute_annual_revenue(plant, sc)
        roi = compute_roi(capex['total_USD'], opex['total_USD'],
                          rev['total'], econ)
        irr_str = f"{roi['IRR']*100:.1f}%" if roi['IRR'] is not None else "N/A"
        pb_str = (f"{roi['payback_years']:.1f}yr"
                  if roi['payback_years'] is not None else ">20yr")
        print(f"  {sc.name:<50s} ${rev['total']:>7.0f} "
              f"${roi['NPV']:>9,.0f} {irr_str:>7s} "
              f"{pb_str:>8s} ${roi['levelized_cost']:>8,.0f}/t")


# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 8: PLOTTING                                                 ║
# ╚═══════════════════════════════════════════════════════════════════════╝

def plot_roi(capex: Dict, opex: Dict, scenarios: List,
             plant: PlantDesign, econ: EconomicParams):
    """Generate comprehensive ROI visualization."""

    fig = plt.figure(figsize=(24, 18))
    fig.suptitle(
        "DAC System — CAPEX / OPEX / ROI Analysis (Pakistan)\n"
        f"Output: {plant.CO2_per_cycle_kg * plant.cycles_per_year:.0f} kg CO₂/yr | "
        f"CAPEX: ${capex['total_USD']:,.0f} | "
        f"OPEX: ${opex['total_USD']:,.0f}/yr",
        fontsize=14, fontweight='bold', y=0.98
    )

    colors_sc = ['#2196F3', '#FF9800', '#4CAF50', '#9C27B0']

    # ── Plot 1: Cumulative Cash Flow (ROI) ───────────────────────────
    ax1 = fig.add_subplot(2, 3, 1)
    for i, sc in enumerate(scenarios):
        rev = compute_annual_revenue(plant, sc)
        roi = compute_roi(capex['total_USD'], opex['total_USD'],
                          rev['total'], econ)
        label = sc.name.split(':')[0]
        ax1.plot(roi['years'], roi['cum_flow'], color=colors_sc[i],
                 lw=2.5, label=label, marker='o', markersize=3)
    ax1.axhline(0, color='black', ls='--', lw=1, alpha=0.5)
    ax1.set_xlabel("Year", fontsize=11)
    ax1.set_ylabel("Cumulative Cash Flow (USD)", fontsize=11)
    ax1.set_title("Return on Investment — Cumulative Cash Flow", fontsize=12)
    ax1.legend(fontsize=8, loc='lower right')
    ax1.grid(True, alpha=0.3)
    ax1.yaxis.set_major_formatter(mticker.FuncFormatter(
        lambda x, p: f"${x:,.0f}"))

    # ── Plot 2: NPV Comparison ───────────────────────────────────────
    ax2 = fig.add_subplot(2, 3, 2)
    sc_names = [sc.name.split(':')[0] for sc in scenarios]
    npvs = []
    for sc in scenarios:
        rev = compute_annual_revenue(plant, sc)
        roi = compute_roi(capex['total_USD'], opex['total_USD'],
                          rev['total'], econ)
        npvs.append(roi['NPV'])
    bar_colors = [c if n > 0 else '#F44336' for c, n in zip(colors_sc, npvs)]
    bars = ax2.bar(sc_names, npvs, color=bar_colors, edgecolor='black',
                   linewidth=0.5)
    ax2.axhline(0, color='black', ls='-', lw=1)
    ax2.set_ylabel("Net Present Value (USD)", fontsize=11)
    ax2.set_title(f"NPV Comparison ({econ.project_life_years}yr, "
                  f"r={econ.discount_rate:.0%})", fontsize=12)
    for bar, val in zip(bars, npvs):
        ax2.text(bar.get_x() + bar.get_width()/2, val,
                 f"${val:,.0f}", ha='center',
                 va='bottom' if val >= 0 else 'top', fontsize=8)
    ax2.grid(True, alpha=0.3, axis='y')
    ax2.yaxis.set_major_formatter(mticker.FuncFormatter(
        lambda x, p: f"${x:,.0f}"))

    # ── Plot 3: CAPEX Breakdown ──────────────────────────────────────
    ax3 = fig.add_subplot(2, 3, 3)
    cat_totals = {}
    for it in capex['items']:
        cat_totals[it.category] = cat_totals.get(it.category, 0) + it.base_cost_USD
    labels_c = list(cat_totals.keys())
    values_c = list(cat_totals.values())
    colors_c = ['#1976D2', '#D32F2F', '#388E3C', '#F57C00']
    wedges, texts, autotexts = ax3.pie(
        values_c, labels=labels_c, autopct='%1.1f%%',
        colors=colors_c[:len(labels_c)], startangle=90,
        textprops={'fontsize': 9}
    )
    ax3.set_title(f"CAPEX Breakdown (${capex['total_USD']:,.0f})", fontsize=12)

    # ── Plot 4: OPEX Breakdown ───────────────────────────────────────
    ax4 = fig.add_subplot(2, 3, 4)
    ocat_totals = {}
    for it in opex['items']:
        ocat_totals[it.category] = ocat_totals.get(it.category, 0) + it.annual_USD
    labels_o = list(ocat_totals.keys())
    values_o = list(ocat_totals.values())
    colors_o = ['#FFC107', '#00BCD4', '#E91E63', '#8BC34A', '#795548']
    wedges2, texts2, autotexts2 = ax4.pie(
        values_o, labels=labels_o, autopct='%1.1f%%',
        colors=colors_o[:len(labels_o)], startangle=90,
        textprops={'fontsize': 9}
    )
    ax4.set_title(f"OPEX Breakdown (${opex['total_USD']:,.0f}/yr)", fontsize=12)

    # ── Plot 5: Revenue Breakdown per Scenario ───────────────────────
    ax5 = fig.add_subplot(2, 3, 5)
    x = np.arange(len(scenarios))
    width = 0.25
    rev_co2_list = []
    rev_h2_list = []
    rev_cr_list = []
    for sc in scenarios:
        rev = compute_annual_revenue(plant, sc)
        rev_co2_list.append(rev['rev_CO2'])
        rev_h2_list.append(rev['rev_H2'])
        rev_cr_list.append(rev['rev_credits'])

    ax5.bar(x - width, rev_co2_list, width, label='CO₂ Sales',
            color='#2196F3', edgecolor='black', linewidth=0.5)
    ax5.bar(x, rev_h2_list, width, label='H₂ Sales',
            color='#4CAF50', edgecolor='black', linewidth=0.5)
    ax5.bar(x + width, rev_cr_list, width, label='Carbon Credits',
            color='#FF9800', edgecolor='black', linewidth=0.5)
    ax5.set_xticks(x)
    ax5.set_xticklabels([sc.name.split(':')[0] for sc in scenarios], fontsize=8)
    ax5.set_ylabel("Revenue (USD/yr)", fontsize=11)
    ax5.set_title("Annual Revenue Breakdown by Scenario", fontsize=12)
    ax5.legend(fontsize=8)
    ax5.grid(True, alpha=0.3, axis='y')
    ax5.yaxis.set_major_formatter(mticker.FuncFormatter(
        lambda x, p: f"${x:,.0f}"))

    # ── Plot 6: Sensitivity — Payback vs CO₂ Price ───────────────────
    ax6 = fig.add_subplot(2, 3, 6)
    co2_prices = np.linspace(100, 8000, 80)
    paybacks = []
    for p in co2_prices:
        rev_test = p * plant.CO2_per_cycle_kg * plant.cycles_per_year / 1000
        rev_test += plant.H2_per_cycle_kg * plant.cycles_per_year * econ.H2_price_USD_kg
        roi_test = compute_roi(capex['total_USD'], opex['total_USD'],
                               rev_test, econ)
        pb = roi_test['payback_years'] if roi_test['payback_years'] is not None else 25
        paybacks.append(min(pb, 25))

    ax6.plot(co2_prices, paybacks, color='#D32F2F', lw=2.5)
    ax6.axhline(econ.project_life_years, color='gray', ls='--', lw=1,
                label=f'Project life ({econ.project_life_years} yr)')
    # Mark scenario prices
    for i, sc in enumerate(scenarios):
        if sc.CO2_price_USD_ton > 0:
            ax6.axvline(sc.CO2_price_USD_ton, color=colors_sc[i], ls=':',
                        lw=1.5, alpha=0.7,
                        label=f'{sc.name.split(":")[0]}: ${sc.CO2_price_USD_ton:.0f}/t')
    ax6.set_xlabel("CO₂ Selling Price (USD/ton)", fontsize=11)
    ax6.set_ylabel("Simple Payback (years)", fontsize=11)
    ax6.set_title("Sensitivity: Payback Period vs CO₂ Price", fontsize=12)
    ax6.set_ylim(0, 25)
    ax6.legend(fontsize=7, loc='upper right')
    ax6.grid(True, alpha=0.3)

    plt.tight_layout(rect=[0, 0.01, 1, 0.94])
    plt.savefig("dac_capex_opex_roi.png", dpi=150, bbox_inches='tight')
    plt.show()
    print("\n  ✓ Plot saved to dac_capex_opex_roi.png")

def run_scaleup_analysis(base_plant: PlantDesign, econ: EconomicParams):
    """Scale-up analysis showing path to profitability."""

    print(f"\n{'='*78}")
    print(f"  SCALE-UP ANALYSIS — Path to Profitability")
    print(f"  Base plant: {base_plant.CO2_per_cycle_kg * base_plant.cycles_per_year:.1f} kg CO₂/yr")
    print(f"  Best market: Food-grade CO₂ Pakistan (Rs.1,200/kg)")
    print(f"{'='*78}")

    scales = [1, 5, 10, 20, 50, 100, 200, 500]
    PKR = econ.USD_PKR

    # ── Base costs at 1× ─────────────────────────────────────────────
    base_capex_absorber = 2227.0
    base_capex_ecstack = 920.0
    base_capex_bop = 1854.0
    base_capex_install = 1542.0
    base_capex = base_capex_absorber + base_capex_ecstack + base_capex_bop + base_capex_install

    base_opex_elec = 79.0
    base_opex_chem = 30.0
    base_opex_replace = 279.0
    base_opex_labor = 690.0
    base_opex_maint = 312.0
    base_opex = base_opex_elec + base_opex_chem + base_opex_replace + base_opex_labor + base_opex_maint

    base_co2_ton = base_plant.CO2_per_cycle_kg * base_plant.cycles_per_year / 1000.0
    base_h2_kg = base_plant.H2_per_cycle_kg * base_plant.cycles_per_year

    # ── Scaling laws ───────────��─────────────────────────────────────
    # Ref: Peters & Timmerhaus, "Plant Design & Economics for Chemical Engineers"
    #   CAPEX: six-tenths rule (exponent 0.6 for process equipment)
    #   Labor: scales with 0.25 power (automation + supervision efficiency)
    #   Replacements: scale with 0.8 (bulk discount on membranes/packing)
    #   Electricity & chemicals: linear (1.0)
    #   Maintenance: 4% of CAPEX

    results = []

    print(f"\n  {'Scale':>5s} {'CO₂':>7s} {'CAPEX':>10s} {'OPEX/yr':>10s} "
          f"{'Rev/yr':>10s} {'Net/yr':>9s} {'LCOC':>9s} {'Payback':>8s} {'NPV(20yr)':>11s}")
    print(f"  {'':>5s} {'t/yr':>7s} {'USD':>10s} {'USD':>10s} "
          f"{'USD':>10s} {'USD':>9s} {'$/t':>9s} {'yr':>8s} {'USD':>11s}")
    print(f"  {'─'*82}")

    for scale in scales:
        # CAPEX scaling
        capex_abs = base_capex_absorber * scale**0.6
        capex_ec = base_capex_ecstack * scale**0.65   # membranes scale slightly better
        capex_bop = base_capex_bop * scale**0.55       # pumps, tanks have good economies
        capex_inst = (capex_abs + capex_ec + capex_bop) * 0.25  # 25% for install
        capex_s = capex_abs + capex_ec + capex_bop + capex_inst

        # OPEX scaling
        opex_elec = base_opex_elec * scale             # linear
        opex_chem = base_opex_chem * scale             # linear
        opex_replace = base_opex_replace * scale**0.8  # bulk discount
        opex_labor = base_opex_labor * scale**0.25     # automation
        opex_maint = 0.04 * capex_s                    # 4% of CAPEX
        opex_insur = 0.01 * capex_s                    # 1% insurance
        opex_s = opex_elec + opex_chem + opex_replace + opex_labor + opex_maint + opex_insur

        # Production
        co2_ton = base_co2_ton * scale
        h2_kg = base_h2_kg * scale

        # Revenue — Scenario A: Pakistan food-grade CO₂ + H₂
        rev_co2 = co2_ton * econ.CO2_food_grade_PKR_kg * 1000 / PKR
        rev_h2 = h2_kg * econ.H2_price_USD_kg
        rev_s = rev_co2 + rev_h2

        # Metrics
        lcoc = (capex_s / econ.project_life_years + opex_s) / co2_ton
        net = rev_s - opex_s

        # NPV (20 yr, with inflation)
        npv = -capex_s
        for yr in range(1, econ.project_life_years + 1):
            opex_yr = opex_s * (1 + econ.inflation_rate)**(yr-1)
            rev_yr = rev_s * (1 + econ.inflation_rate)**(yr-1)
            npv += (rev_yr - opex_yr) / (1 + econ.discount_rate)**yr

        # Payback
        if net > 0:
            # Simple payback (year 1)
            cum = -capex_s
            pb = None
            for yr in range(1, econ.project_life_years + 1):
                opex_yr = opex_s * (1 + econ.inflation_rate)**(yr-1)
                rev_yr = rev_s * (1 + econ.inflation_rate)**(yr-1)
                prev = cum
                cum += rev_yr - opex_yr
                if cum >= 0 and pb is None:
                    frac = -prev / (cum - prev)
                    pb = (yr - 1) + frac
            pb_str = f"{pb:.1f}" if pb is not None else ">20"
        else:
            pb = None
            pb_str = ">20"

        results.append({
            'scale': scale, 'co2_ton': co2_ton, 'h2_kg': h2_kg,
            'capex': capex_s, 'opex': opex_s, 'revenue': rev_s,
            'net': net, 'lcoc': lcoc, 'payback': pb, 'npv': npv,
            'capex_abs': capex_abs, 'capex_ec': capex_ec,
            'capex_bop': capex_bop, 'capex_inst': capex_inst,
            'opex_elec': opex_elec, 'opex_chem': opex_chem,
            'opex_replace': opex_replace, 'opex_labor': opex_labor,
            'opex_maint': opex_maint,
        })

        profitable = "✓" if net > 0 else "✗"
        print(f"  {scale:>4d}× {co2_ton:>7.1f} ${capex_s:>9,.0f} "
              f"${opex_s:>9,.0f} ${rev_s:>9,.0f} ${net:>8,.0f} "
              f"${lcoc:>8,.0f} {pb_str:>7s} ${npv:>10,.0f} {profitable}")

    # ── Find exact breakeven scale ───────────────────────────────────
    print(f"\n  {'─'*78}")
    for i in range(len(results)-1):
        if results[i]['net'] < 0 and results[i+1]['net'] > 0:
            s1 = results[i]['scale']
            s2 = results[i+1]['scale']
            n1 = results[i]['net']
            n2 = results[i+1]['net']
            s_be = s1 + (0 - n1) * (s2 - s1) / (n2 - n1)
            co2_be = base_co2_ton * s_be
            print(f"  ★ BREAKEVEN SCALE: ~{s_be:.0f}× "
                  f"({co2_be*1000:.0f} kg CO₂/yr = {co2_be:.2f} ton/yr)")
            break

    # ══════════════════════════════════════════════════════════════════
    #  PLOTS
    # ══════════════════════════════════════════════════════════════════

    fig, axes = plt.subplots(2, 3, figsize=(24, 14))
    fig.suptitle(
        "DAC Scale-Up Analysis — Pakistan Food-Grade CO₂ Market\n"
        f"Base: {base_co2_ton*1000:.0f} kg/yr | "
        f"Best market: Rs.1,200/kg CO₂ ($4,286/ton) | "
        f"Discount: {econ.discount_rate:.0%} | Inflation: {econ.inflation_rate:.0%}",
        fontsize=14, fontweight='bold', y=0.99
    )

    sc_arr = np.array([r['scale'] for r in results])
    x_labels = [f"{s}×" for s in sc_arr]

    # ── Plot 1: Cumulative Cash Flow for profitable scales ───────────
    ax1 = axes[0, 0]
    colors_roi = ['#F44336', '#FF9800', '#4CAF50', '#2196F3', '#9C27B0']
    plot_scales = [s for s in scales if s >= 5][:5]
    for idx, scale in enumerate(plot_scales):
        r = [x for x in results if x['scale'] == scale][0]
        years = np.arange(0, econ.project_life_years + 1)
        cum = np.zeros(len(years))
        cum[0] = -r['capex']
        for yr in range(1, len(years)):
            opex_yr = r['opex'] * (1 + econ.inflation_rate)**(yr-1)
            rev_yr = r['revenue'] * (1 + econ.inflation_rate)**(yr-1)
            cum[yr] = cum[yr-1] + rev_yr - opex_yr
        c = colors_roi[idx % len(colors_roi)]
        ax1.plot(years, cum, color=c, lw=2.5, marker='o', markersize=3,
                 label=f"{scale}× ({r['co2_ton']:.1f} t/yr)")
    ax1.axhline(0, color='black', ls='--', lw=1.5, alpha=0.7)
    ax1.set_xlabel("Year", fontsize=11)
    ax1.set_ylabel("Cumulative Cash Flow (USD)", fontsize=11)
    ax1.set_title("ROI — Cumulative Cash Flow", fontsize=12)
    ax1.legend(fontsize=8)
    ax1.grid(True, alpha=0.3)
    ax1.yaxis.set_major_formatter(mticker.FuncFormatter(
        lambda x, p: f"${x:,.0f}"))

    # ── Plot 2: NPV vs Scale ────────────────────────────────────────
    ax2 = axes[0, 1]
    npvs = [r['npv'] for r in results]
    bar_colors = ['#4CAF50' if n > 0 else '#F44336' for n in npvs]
    bars = ax2.bar(x_labels, npvs, color=bar_colors, edgecolor='black', lw=0.5)
    ax2.axhline(0, color='black', ls='-', lw=1)
    ax2.set_ylabel("NPV (USD)", fontsize=11)
    ax2.set_title(f"NPV vs Scale ({econ.project_life_years}yr)", fontsize=12)
    for bar, val in zip(bars, npvs):
        va = 'bottom' if val >= 0 else 'top'
        ax2.text(bar.get_x() + bar.get_width()/2, val,
                 f"${val:,.0f}", ha='center', va=va, fontsize=7, rotation=45)
    ax2.grid(True, alpha=0.3, axis='y')
    ax2.yaxis.set_major_formatter(mticker.FuncFormatter(
        lambda x, p: f"${x:,.0f}"))

    # ── Plot 3: LCOC vs Scale ───────────────────────────────────────
    ax3 = axes[0, 2]
    lcocs = [r['lcoc'] for r in results]
    ax3.plot(sc_arr, lcocs, 'o-', color='#D32F2F', lw=2.5, markersize=8)
    ax3.axhline(econ.CO2_food_grade_PKR_kg * 1000 / PKR, color='green',
                ls='--', lw=2, label=f'CO₂ selling price (${econ.CO2_food_grade_PKR_kg*1000/PKR:,.0f}/t)')
    ax3.set_xlabel("Scale Factor", fontsize=11)
    ax3.set_ylabel("Levelized Cost (USD/ton CO₂)", fontsize=11)
    ax3.set_title("Levelized Cost of CO₂ Capture", fontsize=12)
    ax3.set_xscale('log')
    ax3.set_yscale('log')
    ax3.set_xticks(sc_arr)
    ax3.set_xticklabels(x_labels)
    ax3.legend(fontsize=9)
    ax3.grid(True, alpha=0.3, which='both')
    ax3.yaxis.set_major_formatter(mticker.FuncFormatter(
        lambda x, p: f"${x:,.0f}"))

    # ── Plot 4: Revenue vs OPEX ─────────────────────────────────────
    ax4 = axes[1, 0]
    revs = [r['revenue'] for r in results]
    opxs = [r['opex'] for r in results]
    x = np.arange(len(scales))
    w = 0.35
    ax4.bar(x - w/2, revs, w, label='Revenue', color='#4CAF50',
            edgecolor='black', lw=0.5)
    ax4.bar(x + w/2, opxs, w, label='OPEX', color='#F44336',
            edgecolor='black', lw=0.5)
    ax4.set_xticks(x)
    ax4.set_xticklabels(x_labels)
    ax4.set_ylabel("USD/yr", fontsize=11)
    ax4.set_title("Annual Revenue vs OPEX", fontsize=12)
    ax4.legend(fontsize=9)
    ax4.grid(True, alpha=0.3, axis='y')
    ax4.yaxis.set_major_formatter(mticker.FuncFormatter(
        lambda x, p: f"${x:,.0f}"))

    # ── Plot 5: OPEX Breakdown at Scale ─────────────────────────────
    ax5 = axes[1, 1]
    opex_elecs = [r['opex_elec'] for r in results]
    opex_chems = [r['opex_chem'] for r in results]
    opex_repls = [r['opex_replace'] for r in results]
    opex_labrs = [r['opex_labor'] for r in results]
    opex_maints = [r['opex_maint'] for r in results]
    ax5.bar(x, opex_elecs, 0.6, label='Electricity', color='#FFC107')
    ax5.bar(x, opex_chems, 0.6, bottom=np.array(opex_elecs),
            label='Chemicals', color='#00BCD4')
    ax5.bar(x, opex_repls, 0.6,
            bottom=np.array(opex_elecs)+np.array(opex_chems),
            label='Replacement', color='#E91E63')
    ax5.bar(x, opex_labrs, 0.6,
            bottom=np.array(opex_elecs)+np.array(opex_chems)+np.array(opex_repls),
            label='Labor', color='#8BC34A')
    ax5.bar(x, opex_maints, 0.6,
            bottom=np.array(opex_elecs)+np.array(opex_chems)+np.array(opex_repls)+np.array(opex_labrs),
            label='Maintenance', color='#795548')
    ax5.set_xticks(x)
    ax5.set_xticklabels(x_labels)
    ax5.set_ylabel("USD/yr", fontsize=11)
    ax5.set_title("OPEX Breakdown by Scale", fontsize=12)
    ax5.legend(fontsize=7, loc='upper left')
    ax5.grid(True, alpha=0.3, axis='y')
    ax5.yaxis.set_major_formatter(mticker.FuncFormatter(
        lambda x, p: f"${x:,.0f}"))

    # ── Plot 6: Payback Period vs Scale ─────────────────────────────
    ax6 = axes[1, 2]
    pbs = [r['payback'] if r['payback'] is not None else 25 for r in results]
    bar_colors_pb = ['#4CAF50' if p <= econ.project_life_years else '#F44336' for p in pbs]
    ax6.bar(x_labels, pbs, color=bar_colors_pb, edgecolor='black', lw=0.5)
    ax6.axhline(econ.project_life_years, color='gray', ls='--', lw=1.5,
                label=f'Project life ({econ.project_life_years}yr)')
    ax6.set_ylabel("Payback Period (years)", fontsize=11)
    ax6.set_title("Simple Payback Period vs Scale", fontsize=12)
    ax6.legend(fontsize=9)
    ax6.grid(True, alpha=0.3, axis='y')
    for i, pb in enumerate(pbs):
        label = f"{pb:.1f}yr" if pb < 25 else ">20yr"
        ax6.text(i, pb + 0.3, label, ha='center', fontsize=8, fontweight='bold')

    plt.tight_layout(rect=[0, 0.01, 1, 0.94])
    plt.savefig("dac_scaleup_roi.png", dpi=150, bbox_inches='tight')
    plt.show()
    print(f"\n  ✓ Scale-up plot saved to dac_scaleup_roi.png")

    return results
# ╔═══════════════════════════════════════════════════════════════════════╗
# ║  SECTION 9: MAIN EXECUTION                                          ║
# ╚═══════════════════════════════════════════════════════════════════════╝

def run_economic_analysis():
    """Run complete CAPEX/OPEX/ROI analysis."""

    print("\n" + "=" * 78)
    print("  DAC SYSTEM — TECHNO-ECONOMIC ANALYSIS")
    print("  Location: Pakistan | Date: March 2026")
    print("  Product: >99% Pure CO₂ (membrane gas separator)")
    print("=" * 78)

    # Initialize
    econ = EconomicParams()
    plant = PlantDesign()

    print(f"\n  Plant Design Summary:")
    print(f"    Absorber: D={plant.abs_D_m*100:.0f}cm × Z={plant.abs_Z_m:.0f}m | "
          f"η={plant.abs_eta_capture:.1f}%")
    print(f"    EC Cell:  {plant.ec_N_cell}×{plant.ec_A_cell_cm2:.0f}cm² | "
          f"j={plant.ec_j_applied:.1f}A/m² | SEC={plant.SEC_kJ_mol:.0f}kJ/mol")
    print(f"    Cycle:    {plant.cycle_time_h:.1f}h ({plant.cycle_time_h/24:.1f}d) | "
          f"{plant.cycles_per_year:.1f} cycles/yr")
    print(f"    Output:   {plant.CO2_per_cycle_kg * plant.cycles_per_year:.1f} kg CO₂/yr | "
          f"{plant.H2_per_cycle_kg * plant.cycles_per_year:.1f} kg H₂/yr")
    print(f"    Power:    {plant.ec_P_total_W:.1f}W (cell) + "
          f"{plant.blower_power_W:.0f}W (blower) = "
          f"{plant.ec_P_total_W + plant.blower_power_W:.1f}W total")

    # CAPEX
    capex = compute_capex(plant, econ)
    print_capex_report(capex, econ)

    # OPEX
    opex = compute_opex(plant, econ, capex['total_USD'])
    print_opex_report(opex, econ, plant)

    # Revenue scenarios
    scenarios_list = build_revenue_scenarios(econ)
    print_revenue_report(scenarios_list, plant, econ)

    # ROI
    print_roi_summary(capex, opex, scenarios_list, plant, econ)

    # Detailed ROI for best scenario
    print(f"\n{'='*78}")
    print(f"  DETAILED ROI — Scenario A (Food-Grade CO₂, Pakistan)")
     # Detailed ROI for best scenario
    print(f"\n{'='*78}")
    print(f"  DETAILED ROI — Scenario A (Food-Grade CO₂, Pakistan)")
    print(f"{'='*78}")
    rev_A = compute_annual_revenue(plant, scenarios_list[0])
    roi_A = compute_roi(capex['total_USD'], opex['total_USD'],
                        rev_A['total'], econ)
    print(f"  {'Year':>4s} {'Revenue':>10s} {'OPEX':>10s} {'Net':>10s} "
          f"{'Cumulative':>12s} {'NPV':>10s}")
    print(f"  {'─'*60}")
    for yr in range(len(roi_A['years'])):
        print(f"  {roi_A['years'][yr]:>4d} "
              f"${roi_A['revenue_flow'][yr]:>9,.0f} "
              f"${roi_A['opex_flow'][yr]:>9,.0f} "
              f"${roi_A['net_flow'][yr]:>9,.0f} "
              f"${roi_A['cum_flow'][yr]:>11,.0f} "
              f"${roi_A['npv_flow'][yr]:>9,.0f}")

    pb_str = (f"{roi_A['payback_years']:.1f} years"
              if roi_A['payback_years'] is not None else ">20 years")
    irr_str = (f"{roi_A['IRR']*100:.1f}%"
               if roi_A['IRR'] is not None else "N/A")
    print(f"\n  NPV = ${roi_A['NPV']:,.0f}")
    print(f"  IRR = {irr_str}")
    print(f"  Payback = {pb_str}")
    print(f"  Levelized Cost = ${roi_A['levelized_cost']:,.0f}/ton CO₂")
    print(f"{'='*78}")

    # Plot
    plot_roi(capex, opex, scenarios_list, plant, econ)
    # Scale-up analysis
    scaleup = run_scaleup_analysis(plant, econ)

    return capex, opex, scenarios_list, plant, econ, scaleup

if __name__ == "__main__":
    capex, opex, scenarios_list, plant, econ, scaleup = run_economic_analysis()
```
